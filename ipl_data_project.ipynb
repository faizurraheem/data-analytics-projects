# Project 3 — IPL Match Analysis

## Problem Statement
Analyzed IPL match data from 2008 to 2020 to understand 
team performance, toss impact, and player achievements.

## Dataset
- Total Matches: 1095
- Seasons: 2008 to 2020
- Source: Kaggle — IPL Complete Dataset

## Goals
- Identify top 10 winning teams
- Analyze impact of toss on match result
- Understand toss decision preferences
- Find top 10 Man of the Match winners

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
```

```python
df1 = pd.read_csv("matches.csv.csv")
```

```python
w = df1['winner'].value_counts().head(10)
```

```python
w
```

```python
top10 = df1['winner'].value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top10.values, y=top10.index, color='teal')
plt.title("Most wins ipl Franchies (2008-2020)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Number of  wins", fontsize=11, labelpad=10)
plt.ylabel("IPL Team",fontsize=11, labelpad=10)
plt.tight_layout()
plt.show()
```

```python
toss_winner = len(df1[df1['toss_winner']==df1['winner']])
```

```python
toss_winner
```

```python
plt.figure(figsize=(7,5))
decision_colors={"field":"teal" , "bat":"coral"}
sns.countplot(x="toss_decision", data=df1 , hue='toss_decision', palette=decision_colors, legend=False)
plt.title("IPL Toss Dicision Bat / Field strategy", fontsize=15 ,fontweight='bold', pad=15)
plt.xlabel("Dicision After Win the Toss",fontsize=12 , labelpad=11)
plt.ylabel("Number of Matches",fontsize=12 , labelpad=11)
plt.show()
```

```python
player_of_the_match = df1['player_of_match'].value_counts().head(11)
```

```python
player_of_the_match
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=player_of_the_match.values, y=player_of_the_match.index, color='teal')
plt.title('Player with the most "Player of the match" Awards',fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Number of 'Player of the Match' Awards", fontsize=11 , labelpad=10)
plt.ylabel("Player Name", fontsize=11 , labelpad=10)
plt.xticks(range(0, int(max(player_of_the_match.values)) + 3,2))
plt.tight_layout()
plt.show()
```

## Conclusion
- Mumbai Indians were the most successful team with 144 wins
- Toss winner won the match only 50.6% of the time — 
  toss does not guarantee a win
- Most teams preferred to field first after winning the toss (700+)
- AB de Villiers won Man of the Match most times (25 times)
